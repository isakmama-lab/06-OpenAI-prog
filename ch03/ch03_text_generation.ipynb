```python
import openai
import os
from dotenv import load_dotenv

# .env파일의 환경변수를 불러옵니다.
load_dotenv()

# 환경 변수에서 API KEY를 가져옵니다.
api_key = os.getenv("OPENAI_API_KEY")

print(api_key)
```

```python
# API 키 지정하여 클라이언트 선언하기
client = openai.OpenAI(api_key = api_key)
```

### 기본 질문하기

```python
response = client.chat.completions.create(
  model="gpt-5.6-luna",
  messages=[{"role": "user", "content": "피자 만드는 법을 50자로 알려줘."}])

print(response) 
```

```python
print(response.choices[0].message.content)
```

```python
print(response.usage)
```

```python
response.usage.prompt_tokens
```

```python
total_bill_USD = (response.usage.completion_tokens*0.002 +response.usage.prompt_tokens*0.001)/1000
# 26.08.31 기준 환율 1USD = 1370.30 KRW
total_bill_KRW = total_bill_USD*1370.30
print("총 소모 비용 : {} 원".format(total_bill_KRW))
```
