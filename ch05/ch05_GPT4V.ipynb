```python
import os
from openai import OpenAI
from dotenv import load_dotenv

# .env파일의 환경변수를 불러옵니다.
load_dotenv()

# 환경 변수에서 API KEY를 가져옵니다.
api_key = os.getenv("OPENAI_API_KEY")

# GPT-4V와 TTS를 위한 client 선언
client = OpenAI(
    api_key = api_key
)
```

```python
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[
    {
      "role": "user",
      "content": [
        {"type": "text", "text": "이 이미지에 대해서 알려줘"},
        {
          "type": "image_url",
          "image_url": {
            "url": "https://github.com/isakmama-lab/06-OpenAI-prog/blob/main/images.jpg?raw=true",
          },
        },
      ],
    }
  ],
  max_tokens=300,
)

print(response.choices[0])
```
